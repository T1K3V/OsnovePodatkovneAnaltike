# 2. domača naloga

**Naloga**: Izračun plače z upoštevanjem nadur in preverjanjem napačnih vnosov.

### Navodila:
- Uporabnik vnese število ur in urno postavko.
- Če uporabnik opravi več kot 40 ur, se dodatne ure (nadure) obračunajo po 1.5x višji postavki.
- Program uporablja `try/except`, da preveri, ali so vnosi numerični.
- Če vnosi niso številčni (npr. 'deset'), se izpiše opozorilo.


```python
# Program za izračun plače z upoštevanjem nadur in preverjanjem za nenumerične vnose

# Poskusimo dobiti vnos uporabnika z uporabo try / except blokov
try:
    # Uporabnika vprašamo za število ur
    ure = float(input("Vnesi število opravljenih ur: "))

    # Uporabnika vprašamo za urno postavko
    postavka = float(input("Vnesi urno postavko (v €): "))

    # Izračunamo plačo z upoštevanjem nadur (več kot 40 ur)
    if ure <= 40:
        # Če je opravljenih največ 40 ur, ni nadur
        placa = ure * postavka
    else:
        # Če je več kot 40 ur, se prvih 40 ur plača normalno, ostale kot nadure (1.5x)
        navadne_ure = 40
        nadure = ure - 40
        placa = (navadne_ure * postavka) + (nadure * postavka * 1.5)

    # Izpišemo rezultat
    print("Tvoja plača znaša:", round(placa, 2), "€")

except ValueError:
    # Če uporabnik vnese nekaj, kar ni številka (npr. 'deset' namesto 10)
    print("Napaka: Vnesti morate numerične podatke (številke).")
```
